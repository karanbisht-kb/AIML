```python
#from langchain.llms import OpenAI
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_community.tools import WikipediaQueryRun, LLMMathTool
from langchain_community.utilities import WikipediaAPIWrapper

from google.colab import userdata

openai_api_key = userdata.get('openAiKey')
llm = ChatOpenAI(api_key=openai_api_key,temperature=0)
print(llm)

wiki = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper(language="en"))
math_tool = LLMMathTool(llm=llm)
tools = [wiki, math_tool]


#agent = create_agent(
 #   model=llm,
  #  tools=tools,
   # system_prompt="You are a helpful assistant. Use Wikipedia for facts and the math tool for calculations.",verbose=True
#)


agent = create_agent(tools, llm, agent="zero-shot-react-description", verbose=True)
```

```python
print(llm)
```
