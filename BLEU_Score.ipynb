```python
import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

# Download tokenizer (first time only)
nltk.download('punkt')
```

```python
# Reference (ground truth)
reference = "The cat is sitting on the mat".split()

# Candidate (LLM output)
candidate = "The cat is on the mat".split()

# BLEU score with smoothing
smooth = SmoothingFunction().method1
bleu_score = sentence_bleu([reference], candidate, smoothing_function=smooth)

print(f"BLEU Score: {bleu_score:.4f}")
```
